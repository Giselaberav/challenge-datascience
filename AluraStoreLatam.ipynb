### Importación de datos



```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)
```

```python
tienda1['No_tienda'] = "tienda_1"
tienda1.head()
```

```python
tienda2['No_tienda'] = "tienda_2"
tienda2.head()
```

```python
tienda3['No_tienda'] = "tienda_3"
tienda3.head()
```

```python
tienda4['No_tienda'] = "tienda_4"
tienda4.head()
```

```python
#Definir una lista con todos los DataFrames a unir
lista_tiendas = [tienda1, tienda2, tienda3, tienda4]

#  Concatenar los DataFrames
# El parámetro 'axis=0' (por defecto) indica que se unirá de forma vertical (añadiendo filas)
df_all_stores = pd.concat(lista_tiendas, ignore_index=True)
df_all_stores["Ganancia Neta"] = df_all_stores["Precio"] - df_all_stores["Costo de envío"]
```

```python
df_all_stores.head()
```

```python
df_all_stores.tail()
```

#1. Análisis de facturación



```python
# Agrupar por 'No_tienda' y sumar 'Precio'
facturacion_por_tienda = df_all_stores.groupby('No_tienda')['Precio'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=facturacion_por_tienda.index, y=facturacion_por_tienda.values, hue=facturacion_por_tienda.index, palette='coolwarm', legend=False)
plt.title('Facturación Total por Tienda (df_all_stores)')
plt.xlabel('Número de Tienda')
plt.ylabel('Total Facturado ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 4. Productos más y menos vendidos

###Facturación por Producto Tienda 1

```python
# Gráfico 1: Facturación por Producto (Top 10)
# Para no saturar el gráfico, mostramos los 10 productos con mayor facturación
facturacion_por_producto = tienda1.groupby('Producto')['Precio'].sum().sort_values(ascending=False)
top_10_productos_facturacion = facturacion_por_producto.head(10)

plt.figure(figsize=(12, 7))
sns.barplot(x=top_10_productos_facturacion.index, y=top_10_productos_facturacion.values, palette='plasma')
plt.title('Top 10 Productos por Facturación en Tienda 1')
plt.xlabel('Producto')
plt.ylabel('Total Facturado ($)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Gráfico 2: Productos más y menos vendidos (por cantidad de transacciones)
# Primero, obtenemos el conteo de cada producto
conteo_productos = tienda1['Producto'].value_counts()

# Seleccionamos los 5 productos más vendidos y los 5 menos vendidos
top_5_vendidos = conteo_productos.head(5)
bottom_5_vendidos = conteo_productos.tail(5)

# Combinamos los resultados para un solo gráfico
productos_vendidos_comparacion = pd.concat([top_5_vendidos, bottom_5_vendidos])

plt.figure(figsize=(12, 7))
sns.barplot(x=productos_vendidos_comparacion.index, y=productos_vendidos_comparacion.values,
            palette=['green']*len(top_5_vendidos) + ['red']*len(bottom_5_vendidos)) # Diferenciamos con colores
plt.title('Productos Más y Menos Vendidos (por Cantidad de Transacciones) en Tienda 1')
plt.xlabel('Producto')
plt.ylabel('Número de Ventas')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Generar mapa de transacciones en Colombia

### Subtask:
Utilizar la librería `plotly.express` para crear un mapa de dispersión interactivo. Se plotearán las columnas 'lat' y 'lon' de `df_all_stores`, y los puntos serán coloreados según la columna 'No_tienda'. El mapa estará centrado en Colombia para facilitar la visualización.


**Reasoning**:
The subtask requires using `plotly.express` to create an interactive scatter mapbox. I need to import the library first, then create the map with the specified parameters.



```python
import plotly.express as px

# Crea un mapa de dispersión interactivo utilizando px.scatter_mapbox
fig = px.scatter_mapbox(df_all_stores,
                        lat='lat',
                        lon='lon',
                        color='No_tienda',
                        zoom=4.5,
                        center=dict(lat=4.5709, lon=-74.2973),
                        mapbox_style='open-street-map',
                        size_max=15,
                        title='Ubicación de las Transacciones por Tienda en Colombia (Interactive)')

# Muestra el mapa
fig.show()
```

El mapa interactivo de transacciones muestra la distribución geográfica de las ventas de cada tienda en Colombia. Podemos observar que:

*   **Concentración de Ventas:** Todas las tiendas tienen una fuerte concentración de ventas en las principales ciudades de Colombia, como Bogotá, Medellín y Cali, lo que es esperable dada la densidad de población y actividad económica en estas áreas.
*   **Distribución por Tienda:** Aunque hay un solapamiento significativo en las áreas de venta, se pueden identificar patrones sutiles. Por ejemplo, algunas tiendas pueden tener una mayor presencia de transacciones en ciudades específicas o en regiones más al norte o al sur del país.
    *   **Tienda 1 (Naranja):** Parece tener una distribución amplia, con puntos dispersos en varias ciudades principales.
    *   **Tienda 2 (Rojo):** También muestra una distribución en las principales ciudades, quizás con una ligera concentración en Medellín.
    *   **Tienda 3 (Verde):** Presenta una buena cobertura en las ciudades principales, incluyendo algunas transacciones en la costa caribeña (como Santa Marta).
    *   **Tienda 4 (Morado):** Similar a las demás, con transacciones en los centros urbanos más grandes, y también con presencia en ciudades como Cúcuta.
*   **Áreas con Poca o Nula Actividad:** Se aprecian vastas zonas del país sin transacciones registradas, lo que podría indicar áreas de oportunidad para expansión o simplemente zonas rurales con menor densidad de clientes.

En general, el mapa permite una visualización clara de dónde se están realizando las transacciones y cómo cada tienda contribuye a la cobertura geográfica, ayudando a identificar áreas de fuerza y posibles zonas para estrategias de marketing o logística dirigidas.
